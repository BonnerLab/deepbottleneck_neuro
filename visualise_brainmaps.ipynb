```python
import xarray as xr
from pathlib import Path
from helpers.brainmap_helper import convert_array_to_mni, plot_brain_baguette  
from IPython.display import display

# Define your parameters:
subjid = 0                      # Example subject id
layerids = [1, 3, 5, 7, 9, 11]  # List of layer IDs to plot

# Build a dictionary of MNI brain maps.
mni_brain_maps = {}
for layerid in layerids:
    # Construct file paths for backprop and PCA data
    backprop_path = f"/data/apassi1/brainmaps/subj{subjid}_boots_model_{layerid}.nc"
    pca_path = f"/data/apassi1/brainmaps/subj{subjid}_conv_model_{layerid}.nc"
    
    # Open the data arrays from the files
    backprop = xr.open_dataarray(backprop_path)
    pca = xr.open_dataarray(pca_path)
    
    # Compute the difference (this is analogous to your 'd')
    diff = backprop - pca
    
    # Convert the difference to MNI space using the provided function.
    diff_mni = convert_array_to_mni(data=diff, subject=subjid)
    
    # Store it in the dictionary with a label (e.g., "Layer 0", "Layer 1", ...)
    mni_brain_maps[f"Layer {layerid}"] = diff_mni

# Set up a cache directory for storing intermediate PNG images
cache_dir = Path("/tmp/baguette_cache")
cache_dir.mkdir(exist_ok=True, parents=True)
baguette_img = plot_brain_baguette(mni_brain_maps, cache_directory=cache_dir)
# g.figure.set_dpi(300)  # or 200, 300, etc.
display(baguette_img)
```
